Import library pandas dan generate data dummy.

```python
import pandas as pd

data = {'luas' : [6, 8, 10, 14, 18],
        'harga' : [7, 9, 13, 17.5, 18]}
df = pd.DataFrame(data)
df
```

Visualisasikan data tersebut melalui scatterplot dengan menggunakan library matplotlib.

```python
import matplotlib.pyplot as plt

df.plot(kind='scatter', x='luas', y='harga')
plt.title('Perbandingan Luas Tanah dengan Harga Tanah')
plt.xlabel('Luas Tanah (m persegi)')
plt.ylabel('Harga Tanah (dollar)')
plt.xlim(0, 25)
plt.ylim(0, 25)
plt.grid(True)
plt.show()
```

Pisahkan variabel dependen dengan variabel independent.

```python
import numpy as np

X=np.array(df['luas']).reshape(-1, 1)
y=np.array(df['harga'])
```

Latih model regresi linear yang akan diguakan untuk memprediksi harga rumah. Kemudian tampilkan nilai intercept dan slope variablenya.

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print(f'intercept: {model.intercept_}')
print(f'slope: {model.coef_}')
```

Visualisasikan garis persamaan linear yang dihasilkan setelah pelatihan model.

```python
X_vis = np.array([0,25]).reshape(-1,1)
y_vis = model.predict(X_vis)

plt.scatter(X, y)
plt.plot(X_vis, y_vis, color='red')

plt.title('Perbandingan Luas Tanah dengan Harga Tanah')
plt.xlabel('Luas Tanah (m persegi)')
plt.ylabel('HargaTanah (dollar)')
# plt.xlim(0, 25)
# plt.ylim(0, 25)
plt.grid(True)
plt.show()
```

Lakukan prediksi harga rumah untuk luas tanah 12 m^2, 15 m^2, dan 20 m^2.

```python
luas_tanah = np.array([12,15,20]).reshape(-1,1)
luas_tanah

harga_prediksi = model.predict(luas_tanah)
harga_prediksi
```

Tampilkan hasil prediksi harga rumah.

```python
for luas, hrg in zip(luas_tanah, harga_prediksi):
  print(f'Luas Tanah {luas}: harga prediksi: {hrg}')
```
